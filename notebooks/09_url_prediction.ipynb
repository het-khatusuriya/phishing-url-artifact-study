```python
# ============================================================
# URL FEATURE EXTRACTION
# ============================================================

from collections import Counter
from urllib.parse import urlsplit, parse_qs
import ipaddress
import math
import re


# ------------------------------------------------------------
# CONSTANTS
# ------------------------------------------------------------

SUSPICIOUS_TLDS = {
    "zip", "mov", "top", "xyz", "tk", "gq", "ml", "cf",
    "work", "click", "link", "country", "stream", "download",
    "science", "racing", "review", "party", "monster", "cam"
}

URL_SHORTENING_DOMAINS = {
    "bit.ly", "tinyurl.com", "t.co", "goo.gl", "ow.ly",
    "is.gd", "buff.ly", "cutt.ly", "rb.gy", "shorturl.at",
    "tiny.cc", "rebrand.ly", "shorte.st", "adf.ly"
}

SUSPICIOUS_KEYWORDS = {
    "login", "signin", "verify", "verification", "account",
    "secure", "security", "update", "confirm", "password",
    "banking", "wallet", "payment", "invoice", "recover",
    "suspend", "unlock", "validate", "credential"
}

BRAND_NAMES = {
    "google", "paypal", "microsoft", "apple", "amazon",
    "facebook", "instagram", "netflix", "whatsapp", "yahoo",
    "linkedin", "dropbox", "bankofamerica", "chase"
}

EXCESSIVE_HYPHEN_THRESHOLD = 3


# ------------------------------------------------------------
# HELPER FUNCTIONS
# ------------------------------------------------------------

def calculate_entropy(value):
    if not value:
        return 0.0

    counts = Counter(value)
    total = len(value)

    return -sum(
        (count / total) * math.log2(count / total)
        for count in counts.values()
    )


def get_url_parts(url):
    parsed = urlsplit(str(url))

    hostname = parsed.hostname or ""
    path = parsed.path or ""
    query = parsed.query or ""
    scheme = parsed.scheme.lower()

    return parsed, hostname.lower(), path, query, scheme


def is_ip_address(hostname):
    try:
        ipaddress.ip_address(hostname)
        return 1
    except ValueError:
        return 0


def count_letters(value):
    return sum(character.isalpha() for character in value)


def count_digits(value):
    return sum(character.isdigit() for character in value)


def count_query_parameters(query):
    if not query:
        return 0

    return len(parse_qs(query, keep_blank_values=True))


# ------------------------------------------------------------
# MAIN FEATURE EXTRACTION FUNCTION
# ------------------------------------------------------------

def extract_features(url):

    url = str(url)

    parsed, hostname, path, query, scheme = get_url_parts(url)

    url_lower = url.lower()
    hostname_lower = hostname.lower()

    # Basic URL features
    url_length = len(url)
    num_dots = url.count(".")
    num_slashes = url.count("/")
    num_hyphens = url.count("-")
    num_underscores = url.count("_")

    num_special_characters = sum(
        not character.isalnum()
        for character in url
    )

    num_digits = count_digits(url)
    num_letters = count_letters(url)

    digit_to_letter_ratio = (
        num_digits / num_letters
        if num_letters > 0
        else 0.0
    )

    # Path and query features
    path_length = len(path)
    query_string_length = len(query)
    num_query_parameters = count_query_parameters(query)

    # Special-character features
    num_at_symbols = url.count("@")
    num_double_slash_occurrences = url.count("//")

    # Entropy
    url_entropy = calculate_entropy(url)

    # Domain features
    domain_length = len(hostname)

    domain_parts = [
        part for part in hostname.split(".")
        if part
    ]

    num_subdomains = max(len(domain_parts) - 2, 0)

    # TLD
    tld_type = (
        domain_parts[-1]
        if domain_parts and not is_ip_address(hostname)
        else "ip"
    )

    suspicious_tld = int(
        tld_type in SUSPICIOUS_TLDS
    )

    # Punycode / IDN
    punycode_idn_usage = int(
        "xn--" in hostname_lower
    )

    # IP address
    ip_address_instead_of_domain = is_ip_address(
        hostname
    )

    # HTTPS
    https_usage = int(
        scheme == "https"
    )

    # URL shortening
    url_shortening_service = int(
        hostname_lower in URL_SHORTENING_DOMAINS
        or any(
            hostname_lower.endswith("." + domain)
            for domain in URL_SHORTENING_DOMAINS
        )
    )

    # Encoding
    encoded_characters = url.count("%")

    hexadecimal_characters = len(
        re.findall(
            r"(?i)(?:0x[0-9a-f]+|[0-9a-f]{2})",
            url
        )
    )

    # Suspicious keywords
    suspicious_keywords = int(
        any(
            keyword in url_lower
            for keyword in SUSPICIOUS_KEYWORDS
        )
    )

    # Brand names
    brand_name_presence = int(
        any(
            brand in url_lower
            for brand in BRAND_NAMES
        )
    )

    # Excessive hyphens
    excessive_hyphens = int(
        num_hyphens > EXCESSIVE_HYPHEN_THRESHOLD
    )

    # Random-looking domain
    hostname_without_common_prefix = re.sub(
        r"^(www\d?\.)",
        "",
        hostname_lower
    )

    hostname_alphanumeric = re.sub(
        r"[^a-z0-9]",
        "",
        hostname_without_common_prefix
    )

    random_looking_domain = int(
        len(hostname_alphanumeric) >= 12
        and count_digits(hostname_without_common_prefix) >= 3
        and calculate_entropy(hostname_alphanumeric) >= 3.5
    )

    # --------------------------------------------------------
    # RETURN ALL FEATURES
    # --------------------------------------------------------

    return {
        "url_length": url_length,
        "num_dots": num_dots,
        "num_slashes": num_slashes,
        "num_hyphens": num_hyphens,
        "num_underscores": num_underscores,
        "num_special_characters": num_special_characters,
        "num_digits": num_digits,
        "digit_to_letter_ratio": digit_to_letter_ratio,
        "path_length": path_length,
        "query_string_length": query_string_length,
        "num_query_parameters": num_query_parameters,
        "num_at_symbols": num_at_symbols,
        "num_double_slash_occurrences": num_double_slash_occurrences,
        "url_entropy": url_entropy,
        "domain_length": domain_length,
        "num_subdomains": num_subdomains,
        "tld_type": tld_type,
        "suspicious_tld": suspicious_tld,
        "punycode_idn_usage": punycode_idn_usage,
        "ip_address_instead_of_domain": ip_address_instead_of_domain,
        "https_usage": https_usage,
        "url_shortening_service": url_shortening_service,
        "encoded_characters": encoded_characters,
        "hexadecimal_characters": hexadecimal_characters,
        "suspicious_keywords": suspicious_keywords,
        "brand_name_presence": brand_name_presence,
        "excessive_hyphens": excessive_hyphens,
        "random_looking_domain": random_looking_domain
    }


print("✅ URL feature extraction function loaded successfully!")
print("✅ 28 features are ready for new URL prediction.")
```

```python
# ============================================================
# ONE-CELL NEW URL PREDICTION + SHAP EXPLANATION
# ============================================================

from pathlib import Path
from collections import Counter
from urllib.parse import urlsplit, parse_qs
import ipaddress
import math
import re
import warnings

import pandas as pd
import numpy as np
import joblib
import shap


# ============================================================
# 1. FEATURE EXTRACTION
# ============================================================

SUSPICIOUS_TLDS = {
    "zip", "mov", "top", "xyz", "tk", "gq", "ml", "cf",
    "work", "click", "link", "country", "stream", "download",
    "science", "racing", "review", "party", "monster", "cam"
}

URL_SHORTENING_DOMAINS = {
    "bit.ly", "tinyurl.com", "t.co", "goo.gl", "ow.ly",
    "is.gd", "buff.ly", "cutt.ly", "rb.gy", "shorturl.at",
    "tiny.cc", "rebrand.ly", "shorte.st", "adf.ly"
}

SUSPICIOUS_KEYWORDS = {
    "login", "signin", "verify", "verification", "account",
    "secure", "security", "update", "confirm", "password",
    "banking", "wallet", "payment", "invoice", "recover",
    "suspend", "unlock", "validate", "credential"
}

BRAND_NAMES = {
    "google", "paypal", "microsoft", "apple", "amazon",
    "facebook", "instagram", "netflix", "whatsapp", "yahoo",
    "linkedin", "dropbox", "bankofamerica", "chase"
}

EXCESSIVE_HYPHEN_THRESHOLD = 3


def calculate_entropy(value):

    if not value:
        return 0.0

    counts = Counter(value)
    total = len(value)

    return -sum(
        (count / total) * math.log2(count / total)
        for count in counts.values()
    )


def get_url_parts(url):

    parsed = urlsplit(str(url))

    hostname = parsed.hostname or ""
    path = parsed.path or ""
    query = parsed.query or ""
    scheme = parsed.scheme.lower()

    return parsed, hostname.lower(), path, query, scheme


def is_ip_address(hostname):

    try:
        ipaddress.ip_address(hostname)
        return 1
    except ValueError:
        return 0


def count_letters(value):

    return sum(character.isalpha() for character in value)


def count_digits(value):

    return sum(character.isdigit() for character in value)


def count_query_parameters(query):

    if not query:
        return 0

    return len(parse_qs(query, keep_blank_values=True))


def extract_features(url):

    url = str(url)

    parsed, hostname, path, query, scheme = get_url_parts(url)

    url_lower = url.lower()
    hostname_lower = hostname.lower()

    url_length = len(url)
    num_dots = url.count(".")
    num_slashes = url.count("/")
    num_hyphens = url.count("-")
    num_underscores = url.count("_")

    num_special_characters = sum(
        not character.isalnum()
        for character in url
    )

    num_digits = count_digits(url)
    num_letters = count_letters(url)

    digit_to_letter_ratio = (
        num_digits / num_letters
        if num_letters > 0
        else 0.0
    )

    path_length = len(path)
    query_string_length = len(query)

    num_query_parameters = count_query_parameters(query)

    num_at_symbols = url.count("@")
    num_double_slash_occurrences = url.count("//")

    url_entropy = calculate_entropy(url)

    domain_length = len(hostname)

    domain_parts = [
        part for part in hostname.split(".")
        if part
    ]

    num_subdomains = max(
        len(domain_parts) - 2,
        0
    )

    tld_type = (
        domain_parts[-1]
        if domain_parts and not is_ip_address(hostname)
        else "ip"
    )

    suspicious_tld = int(
        tld_type in SUSPICIOUS_TLDS
    )

    punycode_idn_usage = int(
        "xn--" in hostname_lower
    )

    ip_address_instead_of_domain = is_ip_address(
        hostname
    )

    https_usage = int(
        scheme == "https"
    )

    url_shortening_service = int(
        hostname_lower in URL_SHORTENING_DOMAINS
        or any(
            hostname_lower.endswith(
                "." + domain
            )
            for domain in URL_SHORTENING_DOMAINS
        )
    )

    encoded_characters = url.count("%")

    hexadecimal_characters = len(
        re.findall(
            r"(?i)(?:0x[0-9a-f]+|[0-9a-f]{2})",
            url
        )
    )

    suspicious_keywords = int(
        any(
            keyword in url_lower
            for keyword in SUSPICIOUS_KEYWORDS
        )
    )

    brand_name_presence = int(
        any(
            brand in url_lower
            for brand in BRAND_NAMES
        )
    )

    excessive_hyphens = int(
        num_hyphens > EXCESSIVE_HYPHEN_THRESHOLD
    )

    hostname_without_common_prefix = re.sub(
        r"^(www\d?\.)",
        "",
        hostname_lower
    )

    hostname_alphanumeric = re.sub(
        r"[^a-z0-9]",
        "",
        hostname_without_common_prefix
    )

    random_looking_domain = int(
        len(hostname_alphanumeric) >= 12
        and count_digits(
            hostname_without_common_prefix
        ) >= 3
        and calculate_entropy(
            hostname_alphanumeric
        ) >= 3.5
    )

    return {

        "url_length": url_length,
        "num_dots": num_dots,
        "num_slashes": num_slashes,
        "num_hyphens": num_hyphens,
        "num_underscores": num_underscores,
        "num_special_characters": num_special_characters,
        "num_digits": num_digits,
        "digit_to_letter_ratio": digit_to_letter_ratio,
        "path_length": path_length,
        "query_string_length": query_string_length,
        "num_query_parameters": num_query_parameters,
        "num_at_symbols": num_at_symbols,
        "num_double_slash_occurrences": num_double_slash_occurrences,
        "url_entropy": url_entropy,
        "domain_length": domain_length,
        "num_subdomains": num_subdomains,
        "tld_type": tld_type,
        "suspicious_tld": suspicious_tld,
        "punycode_idn_usage": punycode_idn_usage,
        "ip_address_instead_of_domain": ip_address_instead_of_domain,
        "https_usage": https_usage,
        "url_shortening_service": url_shortening_service,
        "encoded_characters": encoded_characters,
        "hexadecimal_characters": hexadecimal_characters,
        "suspicious_keywords": suspicious_keywords,
        "brand_name_presence": brand_name_presence,
        "excessive_hyphens": excessive_hyphens,
        "random_looking_domain": random_looking_domain
    }


# ============================================================
# 2. PROJECT PATH + MODEL
# ============================================================

project_root = Path.cwd().parent

model_path = (
    project_root
    / "models"
    / "weighted_soft_voting_ensemble.joblib"
)

ensemble_model = joblib.load(model_path)


# ============================================================
# 3. LOAD MODEL FEATURE ORDER
# ============================================================

X_test = pd.read_csv(
    project_root
    / "data"
    / "processed"
    / "splits"
    / "X_test.csv"
)

feature_columns = X_test.columns.tolist()


# ============================================================
# 4. ENTER URL
# ============================================================

url = input("Enter URL to test: ").strip()


# ============================================================
# 5. EXTRACT FEATURES
# ============================================================

features = extract_features(url)

new_url_features = pd.DataFrame([features])

new_url_features = new_url_features.reindex(
    columns=feature_columns,
    fill_value=0
)


# ============================================================
# 6. PREDICTION
# ============================================================

prediction = ensemble_model.predict(
    new_url_features
)[0]

probability = ensemble_model.predict_proba(
    new_url_features
)[0]

if prediction == 0:
    label = "LEGITIMATE"
else:
    label = "MALICIOUS"

confidence = max(probability) * 100


# ============================================================
# 7. SHAP FUNCTION
# ============================================================

def shap_predict(data):

    data = pd.DataFrame(
        data,
        columns=feature_columns
    )

    return ensemble_model.predict_proba(data)


# ============================================================
# 8. SHAP EXPLANATION
# ============================================================

print("\nCalculating SHAP explanation...")
print("Please wait...")

background_size = min(
    50,
    len(X_test)
)

background = X_test.sample(
    n=background_size,
    random_state=42
)

explainer = shap.KernelExplainer(
    shap_predict,
    background
)

with warnings.catch_warnings():

    warnings.simplefilter("ignore")

    shap_values = explainer.shap_values(
        new_url_features,
        nsamples=100
    )


# ============================================================
# 9. GET MALICIOUS SHAP VALUES
# ============================================================

if isinstance(shap_values, list):

    malicious_shap = np.asarray(
        shap_values[1]
    )[0]

else:

    shap_array = np.asarray(
        shap_values
    )

    if shap_array.ndim == 3:

        malicious_shap = (
            shap_array[0, :, 1]
        )

    elif shap_array.ndim == 2:

        malicious_shap = (
            shap_array[0]
        )

    else:

        malicious_shap = shap_array


# ============================================================
# 10. TOP FEATURES
# ============================================================

explanation = pd.DataFrame({

    "Feature": feature_columns,

    "Value": new_url_features.iloc[0].values,

    "SHAP Impact": malicious_shap

})

explanation["Absolute Impact"] = (
    explanation["SHAP Impact"].abs()
)

top_features = (
    explanation
    .sort_values(
        "Absolute Impact",
        ascending=False
    )
    .head(5)
)


# ============================================================
# 11. FINAL RESULT
# ============================================================

print("\n")
print("=" * 65)
print("                  NEW URL ANALYSIS")
print("=" * 65)

print(f"\nURL        : {url}")
print(f"Prediction : {label}")
print(f"Confidence : {confidence:.2f}%")

print("\nProbability:")

print(
    f"  Legitimate : "
    f"{probability[0] * 100:.2f}%"
)

print(
    f"  Malicious  : "
    f"{probability[1] * 100:.2f}%"
)

print("\n")
print("-" * 65)
print("       TOP 5 FEATURES AFFECTING THE PREDICTION")
print("-" * 65)

for rank, row in enumerate(
    top_features.itertuples(index=False),
    start=1
):

    feature_name = row[0]
    feature_value = row[1]
    shap_impact = row[2]

    if shap_impact > 0:

        direction = "toward MALICIOUS"

    else:

        direction = "toward LEGITIMATE"

    print(
        f"{rank}. {feature_name}"
        f" | Value: {feature_value}"
        f" | Impact: {shap_impact:+.4f}"
        f" | {direction}"
    )


print("\n")
print("=" * 65)
print("                 ANALYSIS COMPLETE")
print("=" * 65)
```

```python
# ============================================================
# ONE-CELL: NEW URL + PREDICTION + SHAP EXPLANATION
# ============================================================

from pathlib import Path
import sys
import warnings

import pandas as pd
import numpy as np
import joblib
import shap


# ============================================================
# 1. FIND PROJECT ROOT
# ============================================================

project_root = Path.cwd().parent


# ============================================================
# 2. ADD SRC TO PYTHON PATH
# ============================================================

src_path = project_root / "src"

if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))


# ============================================================
# 3. USE YOUR ACTUAL FEATURE EXTRACTOR
# ============================================================

from features.feature_extractor import extract_features


# ============================================================
# 4. LOAD YOUR EXISTING TRAINED ENSEMBLE
# ============================================================

model_path = (
    project_root
    / "models"
    / "weighted_soft_voting_ensemble.joblib"
)

ensemble_model = joblib.load(model_path)


# ============================================================
# 5. LOAD TEST DATA TO GET EXACT FEATURE ORDER
# ============================================================

X_test = pd.read_csv(
    project_root
    / "data"
    / "processed"
    / "splits"
    / "X_test.csv"
)

feature_columns = X_test.columns.tolist()


# ============================================================
# 6. ENTER ANY URL
# ============================================================

url = input("Enter URL to test: ").strip()


# ============================================================
# 7. EXTRACT FEATURES USING YOUR .PY FILE
# ============================================================

features = extract_features(url)

new_url_features = pd.DataFrame([features])


# Make sure feature order matches the trained model
new_url_features = new_url_features.reindex(
    columns=feature_columns,
    fill_value=0
)


# ============================================================
# 8. PREDICTION
# ============================================================

prediction = ensemble_model.predict(
    new_url_features
)[0]


# ============================================================
# 9. PREDICTION PROBABILITY
# ============================================================

probability = ensemble_model.predict_proba(
    new_url_features
)[0]


# ============================================================
# 10. CONVERT LABEL
# ============================================================

if prediction == 0:
    label = "LEGITIMATE"
else:
    label = "MALICIOUS"


confidence = max(probability) * 100


# ============================================================
# 11. SHAP PREDICTION FUNCTION
# ============================================================

def shap_predict(data):

    data = pd.DataFrame(
        data,
        columns=feature_columns
    )

    return ensemble_model.predict_proba(data)


# ============================================================
# 12. CREATE SHAP BACKGROUND DATA
# ============================================================

print("\nCalculating SHAP explanation...")
print("Please wait...")

background = X_test.sample(
    n=min(50, len(X_test)),
    random_state=42
)


# ============================================================
# 13. CREATE SHAP EXPLAINER
# ============================================================

explainer = shap.KernelExplainer(
    shap_predict,
    background
)


# ============================================================
# 14. CALCULATE SHAP VALUES
# ============================================================

with warnings.catch_warnings():
    warnings.simplefilter("ignore")

    shap_values = explainer.shap_values(
        new_url_features,
        nsamples=100
    )


# ============================================================
# 15. GET SHAP VALUES FOR MALICIOUS CLASS
# ============================================================

if isinstance(shap_values, list):

    malicious_shap = np.asarray(
        shap_values[1]
    )[0]

else:

    shap_array = np.asarray(shap_values)

    if shap_array.ndim == 3:

        malicious_shap = shap_array[0, :, 1]

    elif shap_array.ndim == 2:

        malicious_shap = shap_array[0]

    else:

        malicious_shap = shap_array


# ============================================================
# 16. CREATE EXPLANATION TABLE
# ============================================================

explanation = pd.DataFrame({

    "Feature": feature_columns,

    "Value": new_url_features.iloc[0].values,

    "SHAP Impact": malicious_shap

})


explanation["Absolute Impact"] = (
    explanation["SHAP Impact"].abs()
)


# ============================================================
# 17. GET TOP 5 IMPORTANT FEATURES
# ============================================================

top_features = (
    explanation
    .sort_values(
        "Absolute Impact",
        ascending=False
    )
    .head(5)
)


# ============================================================
# 18. DISPLAY FINAL RESULT
# ============================================================

print("\n" + "=" * 65)
print("                  NEW URL ANALYSIS")
print("=" * 65)

print(f"\nURL        : {url}")
print(f"Prediction : {label}")
print(f"Confidence : {confidence:.2f}%")


# ============================================================
# 19. DISPLAY PROBABILITIES
# ============================================================

print("\nProbability:")

print(
    f"  Legitimate : {probability[0] * 100:.2f}%"
)

print(
    f"  Malicious  : {probability[1] * 100:.2f}%"
)


# ============================================================
# 20. DISPLAY TOP SHAP FEATURES
# ============================================================

print("\n" + "-" * 65)
print("       TOP 5 FEATURES AFFECTING THE PREDICTION")
print("-" * 65)


for rank, row in enumerate(
    top_features.itertuples(index=False),
    start=1
):

    feature_name = row[0]

    feature_value = row[1]

    shap_impact = row[2]

    direction = (

        "toward MALICIOUS"

        if shap_impact > 0

        else "toward LEGITIMATE"

    )

    print(
        f"{rank}. {feature_name}"
        f" | Value: {feature_value}"
        f" | Impact: {shap_impact:+.4f}"
        f" | {direction}"
    )


# ============================================================
# 21. COMPLETE
# ============================================================

print("\n" + "=" * 65)
print("                 ANALYSIS COMPLETE")
print("=" * 65)
```

```python
import sys
from pathlib import Path
import joblib
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

project_root = Path.cwd().parent
sys.path.insert(0, str(project_root / "src"))

from features.feature_extractor import extract_features

# Load trained model + training features
model_path = project_root / "models" / "weighted_soft_voting_ensemble.joblib"
ensemble_model = joblib.load(model_path)

split_folder = project_root / "data" / "processed" / "splits"
X_train = pd.read_csv(split_folder / "X_train.csv")
feature_names = X_train.columns.tolist()

print(f"✅ Model loaded! Expected {len(feature_names)} features")

def predict_url(url_string):
    """Predict if URL is malicious."""
    
    print(f"\n{'='*60}")
    print(f"Testing: {url_string}")
    print(f"{'='*60}\n")
    
    try:
        # Extract features
        features_dict = extract_features(url_string)
        
        # Create DataFrame
        url_df = pd.DataFrame([features_dict])
        
        # Align with training features
        # Add missing columns as 0, remove extra columns
        for col in feature_names:
            if col not in url_df.columns:
                url_df[col] = 0
        
        url_df = url_df[feature_names]
        
        # Predict
        prediction = ensemble_model.predict(url_df)[0]
        probability = ensemble_model.predict_proba(url_df)[0]
        
        pred_label = "🚨 MALICIOUS" if prediction == 1 else "✅ LEGITIMATE"
        confidence = max(probability) * 100
        
        print(f"Result: {pred_label}")
        print(f"Confidence: {confidence:.2f}%")
        print(f"Legitimate: {probability[0]:.4f} | Malicious: {probability[1]:.4f}")
        
        return prediction, probability
        
    except Exception as e:
        print(f"❌ Error: {str(e)}")
        return None, None


# Your custom URL
custom = input("\nEnter URL to test (or press Enter to skip): ").strip()
if custom:
    predict_url(custom)

print("\n✅ Done!")
```
